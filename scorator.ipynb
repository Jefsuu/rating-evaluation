```python
import pandas as pd
import requests
from sqlalchemy import create_engine
```

```python
engine = create_engine("postgresql://dev_user:1234@localhost:5432/reviews")
df_max_id = pd.read_sql_query("SELECT coalesce(max(review_id), 0) max_id FROM rated_reviews", con=engine)
max_id = int(df_max_id.iloc[0, 0])
```

```python
df_reviews = pd.read_sql_query(f"SELECT * FROM reviews WHERE id >= {max_id} ORDER BY id ASC LIMIT 100", con=engine)
```

```python
def evaluator_caller(review):
    data = {
		"dataframe_split": {
			"columns": ["review"],
			"data": [review]
		}
	}
    response = requests.post(
		"http://127.0.0.1:5001/invocations",
		json=data,
		# headers={"Content-Type": "application/json"}
        
	)
    print(response.json()['predictions'])
    return response.json()['predictions'][0]
```

```python
data = {
		"dataframe_split": {
			"columns": ["review"],
			"data": [df_reviews.iloc[0]['review']]
		}
	}
```

```python
df_reviews['is_positive'] = df_reviews['review'].apply(evaluator_caller)
```
